# Day 4 — First Look at the Fraud Dataset

This notebook performs an initial exploratory analysis of the fraud transaction dataset. The goal is to understand its structure, missing values, class balance, transaction amounts, and time span before any modeling or feature engineering.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
train_df = pd.read_csv("../data/fraudTrain.csv")
```

```python
train_df.head()
```

```python
print("Shape:", train_df.shape)
print("\nColumns:")
print(train_df.columns.tolist())

print("\nData types:")
print(train_df.dtypes)
```

## Missing Values

```python
missing_count = train_df.isnull().sum()

missing_percentage = (
    missing_count / len(train_df)
) * 100

missing_summary = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percentage": missing_percentage
})

print(missing_summary)
pd.DataFrame({...})
```

## Class Balance

```python
class_counts = train_df["is_fraud"].value_counts()

print(class_counts)
```

```python
fraud_count = (train_df["is_fraud"] == 1).sum()

total_transactions = len(train_df)

fraud_percentage = (
    fraud_count / total_transactions
) * 100

print("Fraud transactions:", fraud_count)
print("Total transactions:", total_transactions)
print("Fraud percentage:", fraud_percentage)
```

## Transaction Amount Distribution

```python
train_df["amt"].describe()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    train_df["amt"],
    bins=50
)

plt.xlabel("Transaction Amount")
plt.ylabel("Number of Transactions")
plt.title("Transaction Amount Distribution")

plt.show()
```

## Fraud vs Non-Fraud Transaction Amount Comparison

```python
fraud_amounts = train_df[
    train_df["is_fraud"] == 1
]["amt"]

legitimate_amounts = train_df[
    train_df["is_fraud"] == 0
]["amt"]
```

```python
print("Fraud transaction amounts:")
print(fraud_amounts.describe())

print("\nLegitimate transaction amounts:")
print(legitimate_amounts.describe())
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    legitimate_amounts,
    bins=50,
    density=True,
    alpha=0.5,
    label="Legitimate"
)

plt.hist(
    fraud_amounts,
    bins=50,
    density=True,
    alpha=0.5,
    label="Fraud"
)

plt.xlabel("Transaction Amount")
plt.ylabel("Density")
plt.title("Transaction Amount: Fraud vs Legitimate")
plt.legend()

plt.show()
```

## Transaction Time Analysis

```python
train_df["trans_date_trans_time"] = pd.to_datetime(
    train_df["trans_date_trans_time"]
)
```

```python
print(train_df["trans_date_trans_time"].dtype)
```

```python
earliest_transaction = train_df[
    "trans_date_trans_time"
].min()

latest_transaction = train_df[
    "trans_date_trans_time"
].max()

print("Earliest transaction:", earliest_transaction)
print("Latest transaction:", latest_transaction)
```

```python
train_df["transaction_hour"] = (
    train_df["trans_date_trans_time"].dt.hour
)
```

```python
fraud_by_hour = train_df[
    train_df["is_fraud"] == 1
]["transaction_hour"].value_counts().sort_index()

print(fraud_by_hour)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    fraud_by_hour.index,
    fraud_by_hour.values
)

plt.xlabel("Hour of Day")
plt.ylabel("Number of Fraud Transactions")
plt.title("Fraud Transactions by Hour of Day")

plt.xticks(range(24))

plt.show()
```

## Initial Observations

1. The dataset is highly imbalanced. Only 7,506 out of 1,296,675 transactions are labelled as fraud, giving a fraud rate of approximately 0.579%. This means accuracy alone could be misleading when evaluating a future fraud model.

2. No missing values were detected across the 23 columns in the training dataset.

3. Transaction amounts are strongly right-skewed. The overall median transaction amount is 47.52, while the maximum reaches 28,948.90, indicating a relatively small number of very large transactions.

4. Fraudulent transactions tend to have substantially higher amounts in this dataset. The median fraudulent transaction amount is 396.51 compared with 47.28 for legitimate transactions.

5. Fraud counts show strong time-of-day clustering. Fraud transactions are particularly concentrated between 22:00 and 03:59, with the highest counts occurring at 22:00 and 23:00. This observation concerns fraud counts rather than fraud rates, so it does not by itself prove that transactions during these hours have a higher probability of fraud.
