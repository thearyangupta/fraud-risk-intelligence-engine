# Day 1 — Temporal Train / Validation / Test Split

Create a chronological train, validation, and test split that preserves the real-world direction of time and prevents future transactions from leaking into earlier datasets.

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/fraudTrain.csv")
```

```python
df["trans_date_trans_time"] = pd.to_datetime(
    df["trans_date_trans_time"]
)
```

```python
df = df.sort_values(
    "trans_date_trans_time"
).reset_index(drop=True)
```

```python
n = len(df)

train_end = int(n * 0.70)
val_end = int(n * 0.85)
```

```python
train = df.iloc[:train_end].copy()

val = df.iloc[
    train_end:val_end
].copy()

test = df.iloc[
    val_end:
].copy()
```

```python
print("Train shape:", train.shape)
print("Validation shape:", val.shape)
print("Test shape:", test.shape)
```

```python
for name, split in [
    ("Train", train),
    ("Validation", val),
    ("Test", test),
]:
    fraud_count = split["is_fraud"].sum()
    fraud_rate = split["is_fraud"].mean()

    print(
        f"{name}: "
        f"rows={len(split):,}, "
        f"fraud_count={fraud_count:,}, "
        f"fraud_rate={fraud_rate:.4%}"
    )
```
