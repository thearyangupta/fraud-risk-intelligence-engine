### Logistic Regression Mechanics

## Mental Model

Logistic regression converts feature values into a fraud probability through:

features → linear score → sigmoid → probability → threshold → prediction

Linear score:

z = w1*x1 + w2*x2 + ... + b

where:

- x = feature value
- w = feature weight
- b = bias/intercept
- z = unbounded linear score

The sigmoid converts z into a value between 0 and 1.

The probability threshold is a separate decision from the model.

```python
import math


def sigmoid(z):
    return 1 / (1 + math.exp(-z))
```

```python
z_values = [-10, -5, 0, 5, 10]

for z in z_values:
    print(
        f"z = {z:>3} -> "
        f"probability = {sigmoid(z):.6f}"
    )
```

```python
w1, w2, b = 1.8, 0.9, -2.0

x1, x2 = 2.1, 0.4

z = w1 * x1 + w2 * x2 + b

p = sigmoid(z)

threshold = 0.5

predict_fraud = p > threshold

print(f"score: {z:.3f}")
print(f"probability: {p:.3f}")
print(f"threshold: {threshold}")
print(f"predict_fraud: {predict_fraud}")
```
